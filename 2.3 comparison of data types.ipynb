##### Comparison of Data Types and Their Methods in Python

#### Integer Numbers (int)
###### Description: Integer numbers (e.g., 1, 2, 100).
###### Mutable: No (Immutable).
###### Methods: No specific methods, but supports arithmetic operations 
###### such as addition, subtraction, multiplication, division, and so on.


---
---

#### Floating-Point Numbers (float)
###### Description: Decimal numbers (e.g., 3.14, 2.71).
###### Mutable: No (Immutable).
###### Methods: No specific methods, but supports arithmetic operations 
###### and mathematical functions (from the math library).


##### *****************************************

#### Strings (str)
###### Description: Text values (e.g., "hello").
###### Mutable: No (Immutable).
###### Methods:
###### Examples include:
###### - capitalize(): Capitalizes the first letter.
###### - lower(): Converts all characters to lowercase.
###### - upper(): Converts all characters to uppercase.
###### - replace(old, new): Replaces occurrences of a substring.
###### - split(delimiter): Splits the string into a list.
###### - strip(): Removes leading and trailing whitespace.
###### Many more methods exist for advanced string manipulation.


##### *****************************************

#### Lists (list)
###### Description: A collection of elements (e.g., [1, 2, 3]).
###### Mutable: Yes.
###### Methods:
###### Examples include:
###### - append(element): Adds an element to the end of the list.
###### - clear(): Removes all elements from the list.
###### - pop(index): Removes and returns the element at the specified index.
###### - sort(): Sorts the list in place.
###### - reverse(): Reverses the order of elements in the list.
###### - remove(element): Removes the first occurrence of the element.


##### *****************************************

#### Tuples (tuple)
###### Description: Immutable collection of elements (e.g., (1, 2, 3)).
###### Mutable: No (Immutable).
###### Methods:
###### - count(element): Counts the occurrences of an element.
###### - index(element): Returns the index of the first occurrence of an element.


##### *********************************************
#### Dictionaries (dict)
###### Description: A collection of key-value pairs (e.g., {"name": "Ahmed", "age": 20}).
###### Mutable: Yes.
###### Methods:
###### Examples include:
###### - keys(): Returns a view of all keys in the dictionary.
###### - values(): Returns a view of all values in the dictionary.
###### - items(): Returns a view of all key-value pairs.
###### - update(new_dict): Updates the dictionary with another dictionary.
###### - pop(key): Removes and returns the value for the given key.
###### - clear(): Removes all key-value pairs from the dictionary.


#### Dictionaries (dict)
###### Description: A collection of key-value pairs (e.g., {"name": "Ahmed", "age": 20}).
###### Mutable: Yes.
###### Methods:
###### Examples include:
###### - keys(): Returns a view of all keys in the dictionary.
###### - values(): Returns a view of all values in the dictionary.
###### - items(): Returns a view of all key-value pairs.
###### - update(new_dict): Updates the dictionary with another dictionary.
###### - pop(key): Removes and returns the value for the given key.
###### - clear(): Removes all key-value pairs from the dictionary.


##### ********************************************

#### Sets (set)
###### Description: Unordered collection of unique elements (e.g., {1, 2, 3}).
###### Mutable: Yes.
###### Methods:
###### Examples include:
###### - add(element): Adds an element to the set.
###### - remove(element): Removes an element from the set (raises an error if the element is not found).
###### - discard(element): Removes an element from the set (does not raise an error if the element is not found).
###### - union(other_set): Returns a set containing all unique elements from both sets.
###### - intersection(other_set): Returns a set of common elements between two sets.
###### - difference(other_set): Returns a set of elements in the first set but not in the second.
###### - isdisjoint(other_set): Checks if two sets have no elements in common.

# Note: 
### Mutable VS Immutable
###### Mutable types (like lists, dictionaries, and sets) can be modified in place,
###### while immutable types (like integers, floats, strings, and tuples) cannot be altered after creation.
### Iterable vs Non-Iterable Objects in Python

###### Iterable objects (like lists, strings, dictionaries, and sets) allow you to loop through their elements one by one,  
###### while non-iterable objects (like integers, floats, and booleans) do not support iteration as they represent single values.
