```python
s = "abcaacbb"

counts = {}

# نعدّ كل حرف
for ch in s:
    if ch in counts:
        counts[ch] += 1
    else:
        counts[ch] = 1

# نعدّ الحروف اللي اتكررت أكتر من مرة
repeated = 0
for value in counts.values():
    if value > 1:
        repeated += 1

print(repeated)
```

## Nested Loop

```python
peoples={
    "ahmed" :{
    "html":"70%",
    "css":"90%",
    "JS":"85%",
    "Python":"95%"
    },
    "sayed":{
    "html":"60%",
    "css":"95%",
    "JS":"85%",
    "Python":"70%"
    },
    "mohamed" :{
    "html":"70%",
    "css":"85%",
    "JS":"80%",
    "Python":"90%"
    }
}
print(peoples["ahmed"])
print(peoples["ahmed"]["css"])

for name in peoples:
    print(f"skills and progress for {name} is: ")
    for skill in peoples [name]:
        print(f"{skill.upper()} => {peoples[name][skill]}")
```

```python
human={
    "name": "Ahmed",
    "age":20,
    "lang":"python"
}

print(human.items())

for info,skill in human.items():
    print(f"{info} => {skill}")
else:
    print("loop is finished")
```
